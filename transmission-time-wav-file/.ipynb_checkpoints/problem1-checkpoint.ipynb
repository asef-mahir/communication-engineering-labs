# Problem 1: Transmission Time of WAV File

```python
import os
```

```python
def transmission_time(wav_file, data_rate):
    """
    Calculate the Transmission time of a file
    """
    if not os.path.isfile(wav_file):
        raise FileNotFoundError(f"File not found : {wav_file}")
    if data_rate <= 0 :
        raise ValueError(f"Data rate must be greater than 0")

    file_size_bytes = os.path.getsize(wav_file)
    message_size_bits = file_size_bytes * 8
    time_seconds = message_size_bits / data_rate
    return time_seconds
```

```python
wav_file = "sample_file.wav"
data_rate = 128000   #128 kbps
```

```python
try:
    file_size = os.path.getsize(wav_file)
    time = transmission_time(wav_file,data_rate)

    print(f"Message size : {file_size :,} bytes")           # formatting using :, this separates the big number using commas
    print(f"Message size: {file_size * 8 :,} bits")       
    print(f"Data rate : {data_rate :,} kbps")
    print(f"Transmission time : {time :.4f} seconds")       # 4 digits after the decimal point
    
except(FileNotFoundError, ValueError) as e:
    print(f"Error: {e}")
```
