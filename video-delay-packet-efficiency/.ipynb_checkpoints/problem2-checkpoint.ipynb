# Problem 2: Transmission + Propagation Delay with Packets

```python
# User inputs
video_size_mb = float(input("Enter video size (MB):"))
data_rate_mbps = float(input("Enter data rate (Mbps):"))
distance_km = float(input("Enter distance (km):"))
propagation_speed = float(input("Enter propagation speed (m/s):"))
```

```python
# Unit Conversion
video_size_bytes = video_size_mb * 1_000_000
video_size_bits = video_size_bytes * 8

data_rate_bps = data_rate_mbps * 1_000_000
distance_m = distance_km * 1_000
```

```python
# 1. Transmission delay
transmission_delay = video_size_bits / data_rate_bps
# 2. Propagation delay
propagation_delay = distance_m / propagation_speed
# 3. Total delay
total_delay = transmission_delay + propagation_delay
```

```python
# Packet Information
payload_size = 1_000       # bytes
header_size = 100          # bytes
packet_size = payload_size + header_size

# Calculate number of packets
number_of_packets = video_size_bytes / payload_size

# Total transmitted bytes including headers
total_transmitted_bytes = number_of_packets * packet_size

# Convert total transmitted data to bits
total_transmitted_bits = total_transmitted_bytes * 8
```

```python
# 4. Effective transmission time
effective_transmission_time = total_transmitted_bits / data_rate_bps

# 5. Transmission efficiency
transmission_efficiency = (payload_size / packet_size) * 100
```

```python
print("\n========== RESULTS ==========")

print(f"Video size: {video_size_mb:.2f} MB")
print(f"Data rate: {data_rate_mbps:.2f} Mbps")
print(f"Distance: {distance_km:.2f} km")
print(f"Propagation speed: {propagation_speed:.2e} m/s")      # scientific format: 2.65e+7

print(f"\nTransmission delay: {transmission_delay:.4f} seconds")
print(f"Propagation delay: {propagation_delay:.6f} seconds")
print(f"Total delay: {total_delay:.4f} seconds")

print(f"\nPayload size: {payload_size} bytes")
print(f"Header size: {header_size} bytes")
print(f"Total packet size: {packet_size} bytes")
print(f"Number of packets: {number_of_packets:.0f}")       # .0f rounds to the nearest integer

print(f"Effective transmission time: {effective_transmission_time:.4f} seconds")
print(f"Transmission efficiency: {transmission_efficiency:.2f}%")
```
