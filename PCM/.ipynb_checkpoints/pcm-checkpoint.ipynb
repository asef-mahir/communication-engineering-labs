```python
import numpy as np
```

```python
# Given data

sample_values = np.array([-5, -3, -1, 1, 3, 5, 3, 1])

bits_per_sample = 3

min_value = -5
max_value = 5

sampling_rate = 8000
```

```python
# Number of quantization levels

levels = 2 ** bits_per_sample

print("Number of quantization levels:", levels)
```

```python
# Quantization step

step_size = (max_value - min_value) / levels

print("Quantization step size:", step_size, "V")
```

```python
# Quantize the samples (using mid-rise quantization)

quantized_values = min_value + (
    np.floor((sample_values - min_value) / step_size) + 0.5
) * step_size

print("Original samples:")
print(sample_values)

print("\nQuantized samples:")
print(quantized_values)
```

```python
# Quantization indices

quantization_indices = np.floor(
    (sample_values - min_value) / step_size
).astype(int)

print("Quantization indices:")
print(quantization_indices)
```

```python
# Convert to Binary

binary_values = [
    format(index, f'0{bits_per_sample}b')          # format(5, '03b') = 101
    for index in quantization_indices
]

print("Binary PCM codes:")

for sample, index, binary in zip(
    sample_values,
    quantization_indices,
    binary_values
):
    print(sample, "->", index, "->", binary)
```

```python
# PCM Bit Rate
bit_rate = sampling_rate * bits_per_sample

print("PCM bit rate:", bit_rate, "bits/s")
print("PCM bit rate:", bit_rate / 1000, "kbps")
```
