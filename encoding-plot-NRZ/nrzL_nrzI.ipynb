```python
import matplotlib.pyplot as plt

bits = "10110010"
```

```python
# NRZ-L encoding function
def nrz_l(bits):
    signal = []
    for bit in bits:
        if bit == "0":
            signal.append(1)
        else:
            signal.append(-1)
    return signal
```

```python
# NRZ-I encoding function
def nrz_i(bits):
    signal = []
    level = 1
    for bit in bits:
        if bit == "1":
            level = -level
        signal.append(level)
    return signal
```

```python
# Generating signals and printing
nrz_l_signal = nrz_l(bits)
nrz_i_signal = nrz_i(bits)

print("Bit Stream :", bits)
print("NRZ-L      :", nrz_l_signal)
print("NRZ-I      :", nrz_i_signal)
```

```python
# Plotting function
def plot_signal(ax, signal, title):
    
    t = []
    for i in range(len(signal)):
        t.extend([i, i+1])

    y = []
    for value in signal:
        y.extend([value, value])

    ax.plot(t, y, drawstyle='steps-post', linewidth=2)

    # Bit boundaries
    for i in range(len(bits) + 1):
        ax.axvline(i, color='gray', linestyle='--', alpha=0.4)

    # Bit labels
    for i,bit in enumerate(bits):
        ax.text(i+0.5, 1.3, bit, ha='center', fontsize=12, fontweight='bold')

    ax.set_title(title)
    ax.set_xlim(0, len(bits))
    ax.set_ylim(-1.5, 1.5)
    ax.set_yticks([-1,1])
    ax.set_yticklabels(["Low", "High"])
    ax.grid(axis='y', alpha=0.3)
```

```python
# Final plot creation
fig , axes = plt.subplots(2, 1, figsize=(12,6))

plot_signal(axes[0], nrz_l_signal, "NRZ-L Encoding")
plot_signal(axes[1], nrz_i_signal, "NRZ-I Encoding")

axes[1].set_xlabel("Bit Time")

plt.tight_layout()
plt.show()
```
