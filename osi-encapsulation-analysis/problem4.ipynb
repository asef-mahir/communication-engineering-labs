# Problem 4: OSI Encapsulation

```python
# Application data and headers
application_data = 500

transport_header = 20
network_header = 20
data_link_header = 14
data_link_trailer = 4
```

```python
# Adding headers step by step
transport_data = application_data + transport_header
network_packet = transport_data + network_header
data_link_frame = network_packet + data_link_header + data_link_trailer
```

```python
print("--- OSI Encapsulation ---")
print(f"Application Layer : {application_data} bytes")
print(f"Transport Layer   : {transport_data} bytes")
print(f"Network Layer     : {network_packet} bytes")
print(f"Data Link Layer   : {data_link_frame} bytes")
```

```python
# Calculate overhead and efficiency
total_overhead = data_link_frame - application_data
efficiency = (application_data / data_link_frame) * 100

print("Total overhead:", total_overhead, "bytes")
print(f"Efficiency: {efficiency:.2f}%")
```
